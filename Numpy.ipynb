#Numpy Library
Numpy stands for numerical python.

It is a fundamental open source library for scientific computing

Its core feature is N - Dimensional array object, also known as ndarray.

**N- Dimensional Array(ndarray):**
  This is central data structure in numpy, allowing for efficient storage and manipulation of numerical datqa in various dimensions.

**High Performance:**
  Numpy arrays are usually more faster than python list, due to its underlying implementation in C and Fortran.

**Mathematical Functions:**
  It offers a vast collection of high - level - mathematical functions design to operate on these arrays, including linear algebra, fourier transform, random number geneeration, and various statistical operation.

**Broadcasting:**
  NumnPy's broadcasting mechanism allows operations on arrays with different shapes, provided they are compatible, simplifying code and improving efficiency

**Foundation for Other Libraries:**
  NumPy serves as a foundation for many other popular libraries in  the python data science ecosystem such as SciPy, Pandas, matplotlib.
